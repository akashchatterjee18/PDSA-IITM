# Computing Prime Numbers

### Finding Factors

```python
def factors(n):
  factorlist = []
  for i in range(1, n+1):
    if n % i == 0:
      factorlist.append(i)
  return factorlist
```

### Prime check

```python
def prime(n):
  return(factors(n) == [1, n])
```

### Alternately



```python
def prime(n):
  return(len(factors(n)) == 2)
```

### Finally

```python
primelist = []
for i in range(1,101):
  if (prime(i)):
    primelist.append(i)
print(primelist)
```

### First n primes

```python
def firstmprimes(m):
  (count,i,pl) = (0,1,[])
  while (count < m):
    if (prime(i)):
      (count,pl) = (count+1, pl+[i])
    i = i+1
  return pl
```

```python
m=100
print(firstmprimes(m))
```

### Counting Differences Between Consecutive Prime Numbers

```python
def primediffs(n):
    lastprime = 2
    pd = {}  # Dictionary for prime differences

    for i in range(3, n + 1):
        if prime(i):
            d = i - lastprime
            lastprime = i

            if d in pd.keys():
                pd[d] = pd[d] + 1
            else:
                pd[d] = 1

    return pd
```

```python
primediffs(500)
```

# Plotting sin and cos graph

# Computing GCD/HCF

```python
def gcd(m,n):
  for i in range(1, min(m,n)+1):
    if (m%i == 0) and (n%i == 0):
      hcf = i
  return(hcf)
```

```python
gcd(150,640)
```

### Alternatively using the differences of numbers (using Recursion)

```python
def hcf(m,n):
  a,b = max(m,n), min(m,n)
  if a%b == 0:
    return b
  else:
    return hcf(b, a-b)
```

```python
hcf(150,650)
```

### Using Euclid's Algorithm

```python
def hcf1(m,n):
  a,b = max(m,n), min(m,n)
  if a%b == 0:
    return b
  else:
    return hcf1(b, a%b)
```

```python
hcf1(160,920)
```
